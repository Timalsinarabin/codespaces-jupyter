<h1>Chapter 1: Text Preprocessing</h1>

```python
import os
import pandas as pd
import re
import string
from textblob import TextBlob
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
import emoji
```

```python
df = pd.read_csv("IMDB.csv")
df = df.sample(
    n=5000,
    random_state=42
).reset_index(drop=True)
```

```python

# remove html tags
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)

# remove urls
def remove_url(text):
    pattern = re.compile(r'https??://\S+|www\.\S+')
    return pattern.sub(r'',text)

# emoji_remove
def emoji_remove(text):
    return emoji.replace_emoji(text, replace='')

# tokenzie
def tokenize(text):
    return word_tokenize(text)

# punctuation removal
exclude = string.punctuation
def rem_punctuation(text):
    return text.translate(str.maketrans('','',exclude))

#remove stopwords
stop_words = set(stopwords.words('english'))
def remove_stopwords(tokens):
    return [
        word for word in tokens
        if word not in stop_words
        ]

# remove spacing
def clean_text(text):
    return re.sub(r'\s+', ' ',text).strip()

# Lower case
df["clean_review"] = df["review"].str.lower()

# correct spelling
def correctspell(text):
    return str(TextBlob(text).correct())
```

```python
df["clean_review"] = df["clean_review"].apply(remove_html_tags)

df["clean_review"] = df["clean_review"].apply(remove_url)

df["clean_review"] = df["clean_review"].apply(emoji_remove)

df["clean_review"] = df["clean_review"].apply(rem_punctuation)

df["clean_review"] = df["clean_review"].apply(clean_text)
                                              
df["token"] = df["clean_review"].apply(tokenize)

df["token"] = df["token"].apply(remove_stopwords)
```
