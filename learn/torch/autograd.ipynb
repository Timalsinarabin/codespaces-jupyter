```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
z = w*x + b
z
```

```python
y_pred = torch.sigmoid(z)
y_pred
```

```python
loss_fn = torch.nn.BCELoss()
loss = loss_fn(y_pred, y)
loss
```

```python
loss.backward()
```

```python
w.grad
```

```python
b.grad
```

```python

# Input and target
x = torch.tensor(2.0)
y = torch.tensor(4.0)

# Parameters of the single neuron
w = torch.tensor(1.5, requires_grad=True)
b = torch.tensor(0.51, requires_grad=True)

# Forward pass
y_pred = w * x + b

# Mean squared error
loss = (y_pred - y) ** 2

# Backward pass
loss.backward()

print("Prediction:", y_pred.item())
print("Loss:", loss.item())
print("dL/dw:", w.grad.item())
print("dL/db:", b.grad.item())
```
